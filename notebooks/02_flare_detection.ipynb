```python
from pathlib import Path
import sys
import pandas as pd

PROJECT_ROOT = Path.cwd().resolve()
if not (PROJECT_ROOT / 'scripts').exists(): PROJECT_ROOT = PROJECT_ROOT.parent
sys.path.insert(0, str(PROJECT_ROOT / 'scripts'))
from load_solexs import read_light_curve
from detect_flares import detect_peaks
from utils import CATALOG_DIR, ensure_output_directories, solexs_light_curves
```

```python
files = solexs_light_curves()
if not files:
    raise FileNotFoundError('Add a light curve under data/solexs/YYYYMMDD/ first.')
frame = read_light_curve(files[0])
frame.columns
```

```python
# Replace this with the correct counts column after inspecting the data.
COUNT_COLUMN = 'counts'
peaks = detect_peaks(frame[COUNT_COLUMN])
peaks['source_file'] = files[0].name
ensure_output_directories()
peaks.to_csv(CATALOG_DIR / 'flare_events.csv', index=False)
peaks.head()
```
